```python
# Tinker smoke-test submission: Kaggle auto-unzips uploaded archives, so the
# dataset contains adapter_config.json + adapter_model.safetensors as flat
# files. We re-zip them as /kaggle/working/submission.zip so the competition
# metric kernel can unzip and load via vLLM LoRARequest. If this loads
# without error we have confirmed Tinker's target_modules=all-linear adapter
# is vLLM-compatible end to end.
import glob, os, zipfile
src_dir = os.path.dirname(next(iter(glob.glob('/kaggle/input/**/adapter_config.json', recursive=True))))
print('Adapter source dir:', src_dir)
out = '/kaggle/working/submission.zip'
with zipfile.ZipFile(out, 'w', zipfile.ZIP_DEFLATED) as zf:
    for fname in ['adapter_config.json', 'adapter_model.safetensors']:
        p = os.path.join(src_dir, fname)
        assert os.path.exists(p), p
        zf.write(p, arcname=fname)
print('Wrote', out, '=', os.path.getsize(out), 'bytes')
```
